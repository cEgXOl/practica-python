La asignación conserva la referencia al mismo objeto; `list()` y `[:]` crean listas independientes. La rebaja se aplica a una copia para preservar los precios originales.

```python
precios_originales = [59.90, 149.00, 219.00, 18.50, 189.00, 279.00]
precios_rebajados = precios_originales
precios_rebajados[0] = 49.90
print("Tras asignar y cambiar:", precios_originales, precios_rebajados)
precios_originales[0] = 59.90  # Se restaura el objeto compartido.
copia_list = list(precios_originales)
copia_slice = precios_originales[:]
copia_list[0] = 49.90
print("Original y copias independientes:", precios_originales, copia_list, copia_slice)
for i in range(len(copia_slice)):
    copia_slice[i] = round(copia_slice[i] * 0.90, 2)
print("Copia con 10 % de rebaja:", copia_slice)
copia_slice.extend([39.90, 79.90])
copia_slice.insert(2, 24.90)
del copia_slice[4]
descartado = copia_slice.pop()
print("Tras altas y bajas:", copia_slice, "| descartado:", descartado)
print("IDs distintos:", id(precios_originales), id(copia_list), id(copia_slice))
print("Identidades independientes:", precios_originales is not copia_list, precios_originales is not copia_slice)
```

## Ejercicio 5 — Funciones propias para el cálculo de importes

El importe neto descuenta un porcentaje del bruto. `resumen_pedido()` devuelve las tres magnitudes relacionadas en una tupla para poder desempaquetarlas.

```python
def calcular_importe(unidades, precio_unitario, descuento_pct=0):
    """Devuelve el importe neto redondeado de una línea de pedido."""
    bruto = unidades * precio_unitario
    return round(bruto * (1 - descuento_pct / 100), 2)

def resumen_pedido(unidades, precio_unitario, descuento_pct=0):
    """Devuelve (importe bruto, ahorro, importe neto) de un pedido."""
    bruto = round(unidades * precio_unitario, 2)
    neto = calcular_importe(unidades, precio_unitario, descuento_pct)
    return bruto, round(bruto - neto, 2), neto

bruto, ahorro, neto = resumen_pedido(unidades=3, precio_unitario=59.90, descuento_pct=10)
print("Resumen por palabras clave:", bruto, ahorro, neto)
help(calcular_importe)
pedidos = [(3, 59.90, 10), (1, 279.00, 0), (5, 18.50, 20),
           (2, 149.00, 5), (4, 89.90, 15)]
importes_pedidos = [calcular_importe(*pedido) for pedido in pedidos]
print("Importes:", importes_pedidos)
print("Máximo:", max(importes_pedidos), "| mínimo:", min(importes_pedidos))
print("Ordenados:", sorted(importes_pedidos), "| cantidad:", len(importes_pedidos))
```

## Ejercicio 11 — Estadística descriptiva sobre telemetría simulada

El generador con semilla fija permite repetir exactamente esta simulación. La correlación mide la relación lineal entre las dos magnitudes generadas.

```python
rng = np.random.default_rng(2024)
temperatura = rng.normal(68, 6.5, 500)
vibracion = np.abs(rng.normal(2.4, 0.8, 500))
media = temperatura.mean()
mediana = np.median(temperatura)
desviacion = temperatura.std()
print("Temperatura: media", round(media, 3), "mediana", round(mediana, 3),
      "desviación", round(desviacion, 3), "mín", round(temperatura.min(), 3),
      "máx", round(temperatura.max(), 3))
print("Percentiles vibración 25, 50, 75, 95:", np.round(np.percentile(vibracion, [25, 50, 75, 95]), 3))
telemetria = np.column_stack((temperatura, vibracion))
print("Forma de telemetría:", telemetria.shape)
correlacion = np.corrcoef(telemetria, rowvar=False)
print("Matriz de correlación:\n", np.round(correlacion, 3))
umbral = media + 2 * desviacion
superan = (temperatura > umbral).sum()
print("Umbral:", round(umbral, 3), "| lecturas que lo superan:", superan,
      "| porcentaje:", round(superan / len(temperatura) * 100, 2), "%")
```

La media y la mediana próximas indican que no hay una asimetría marcada en esta muestra. La correlación observada es próxima a cero, por lo que no se aprecia una relación lineal relevante.

La semilla fija hace que un tercero genere los mismos valores con el mismo generador y versiones compatibles. Así puede verificar cada estadístico y detectar diferencias causadas por cambios en el análisis.

```python
from pathlib import Path

ventas["importe_bruto"] = ventas["unidades"] * ventas["precio_unitario"]
ventas["importe_neto"] = (ventas["importe_bruto"] * (1 - ventas["descuento_pct"] / 100)).round(2)
ventas["ticket_medio"] = ventas["importe_neto"] / ventas["unidades"]
facturacion_region = ventas.groupby("region")["importe_neto"].sum().sort_values(ascending=False)
categoria_canal = ventas.groupby(["categoria", "canal"]).agg({"importe_neto": "mean", "unidades": "sum"})
top_productos = ventas.groupby("producto")["importe_neto"].sum().nlargest(5)
print("Facturación total:", round(ventas["importe_neto"].sum(), 2), "€")
print("Facturación por región:\n", facturacion_region)
print("Categoría y canal:\n", categoria_canal)
print("Top 5 productos:\n", top_productos)
ventas["satisfaccion"] = ventas["satisfaccion"].fillna(ventas["satisfaccion"].median())
ventas["canal"] = ventas["canal"].fillna("Desconocido")
Path("../outputs").mkdir(exist_ok=True)
facturacion_region.to_csv("../outputs/facturacion_por_region.csv", header=["importe_neto"])
print("Nulos tras completar:\n", ventas[["satisfaccion", "canal"]].isna().sum())
print("Exportado: ../outputs/facturacion_por_region.csv")
```

## Ejercicio 16 — Operadores de comparación y booleanos

En expresiones escalares se usan `and`, `or` y `not`. Para comparar arrays elemento a elemento se usa `np.logical_and()` o `&` con cada comparación entre paréntesis.

```python
expresiones = [2 < 3, "Norte" == "norte", True + True + False,
               3 != 3.0, "Zapatilla" > "Mochila"]
for expresion in expresiones:
    print(expresion)
unidades = 6
precio = 149.0
descuento = 15
pedido_volumen = unidades > 4 and descuento >= 10
precio_extremo = precio < 50 or precio > 200
print("Pedido de volumen con descuento:", pedido_volumen)
print("Precio fuera de 50 a 200:", precio_extremo)
print("No cumple la condición de volumen y descuento:", not pedido_volumen)
temp = np.array([64.2, 88.1, 71.0, 92.5, 69.8])
vib = np.array([1.9, 3.4, 2.1, 4.0, 2.2])
try:
    temp > 70 and vib > 3
except ValueError as exc:
    print("Error con and:", type(exc).__name__, "—", exc)
filtro_logical = np.logical_and(temp > 70, vib > 3)
filtro_amp = (temp > 70) & (vib > 3)
print("np.logical_and:", filtro_logical)
print("Operador &:     ", filtro_amp)
print("Resultados iguales:", np.array_equal(filtro_logical, filtro_amp))
```

`2 < 3` es verdadero porque 2 es menor que 3. `"Norte" == "norte"` es falso porque las mayúsculas cuentan. `True + True + False` vale 2 porque los booleanos se suman como 1 y 0. `3 != 3.0` es falso porque ambos números tienen el mismo valor. `"Zapatilla" > "Mochila"` es verdadero por el orden lexicográfico de sus letras iniciales.

El `not` indica que este pedido sí cumple la condición conjunta de volumen y descuento. En arrays, `and` intenta obtener un único valor de verdad para varios elementos y por eso lanza `ValueError`.

## Ejercicio 17 — `if` / `elif` / `else` en mantenimiento predictivo

Se prueban primero los umbrales más altos para que cada temperatura caiga en una sola categoría. La recomendación final se basa en las frecuencias calculadas sobre `sensores`.

```python
def nivel_riesgo(temperatura):
    if temperatura > 85:
        return "CRÍTICO"
    elif temperatura > 76:
        return "ALTO"
    elif temperatura > 70:
        return "MEDIO"
    else:
        return "BAJO"

for valor in [92.5, 80.0, 72.3, 64.1]:
    print(valor, "->", nivel_riesgo(valor))
sensores["riesgo"] = sensores["temperatura_c"].apply(nivel_riesgo)
print("Primeras lecturas:\n", sensores[["id_maquina", "temperatura_c", "riesgo"]].head())
print("Riesgos:\n", sensores["riesgo"].value_counts())
criticos = sensores[sensores["riesgo"] == "CRÍTICO"]
print("Críticos por máquina:\n", criticos["id_maquina"].value_counts())
print("Estados originales:\n", sensores["estado"].value_counts())
```

```python
online_500 = ventas[(ventas["canal"] == "Online") & (ventas["importe_neto"] > 500)]
norte_centro = ventas[ventas["region"].isin(["Norte", "Centro"])]
descuento_10_20 = ventas[ventas["descuento_pct"].between(10, 20)]
electronica = ventas[(ventas["categoria"] == "Electronica") &
                     (ventas["unidades"] > 2) & (ventas["descuento_pct"] != 0)]
no_oficina = ventas[~(ventas["categoria"] == "Oficina")]
online_500_query = ventas.query('canal == "Online" and importe_neto > 500')
print("Online y >500:", len(online_500), "| Norte/Centro:", len(norte_centro))
print("Descuento 10-20:", len(descuento_10_20), "| Electrónica, >2 y descuento:", len(electronica))
print("No Oficina:", len(no_oficina), "| query equivale:", online_500.equals(online_500_query))
senior_remoto = empleados[(empleados["departamento"] == "Ingenieria") &
                          (empleados["modalidad"] == "Remoto") &
                          (empleados["antiguedad_anios"] > 5)].sort_values("salario_base", ascending=False)
print("Ingeniería, remoto y >5 años:\n", senior_remoto)
senior_remoto.to_csv("../outputs/ingenieria_remoto_senior.csv")
print("Exportado: ../outputs/ingenieria_remoto_senior.csv")
```

`.query()` deja la condición en una cadena compacta y puede resultar más legible; las máscaras son útiles cuando se reutilizan o combinan filtros.

## Ejercicio 19 — Bucles sobre estructuras de datos

El `while` repone hasta llegar al objetivo; los distintos `for` recorren secuencias, filas y pares clave-valor según la estructura.

```python
stock = 12
lote = 0
while stock < 60:
    lote = lote + 1
    stock = stock + 8
    print("Lote", lote, "-> stock:", stock)
regiones = ["Norte", "Sur", "Este", "Oeste", "Centro"]
for region in regiones:
    print(region.upper())
for numero, region in enumerate(regiones, start=1):
    print(f"Región {numero}: {region}")
for almacen, producto, unidades_stock, coste in inventario:
    print(almacen, "|", producto, "|", round(unidades_stock * coste, 2), "€")
for region, tarifa in tarifas_envio.items():
    print(region, "->", f"{tarifa:.2f} €")
for precio_unitario in np_precios:
    print("Precio con recargo:", round(precio_unitario * 1.04, 2))
print("Stock final:", stock)
```

Se usa `.items()` para obtener en cada vuelta la clave y su valor; sin él, el bucle entrega únicamente las claves del diccionario.
